```python
#library loading
import torch
from torch import nn
import numpy as np
import pandas as pd

#data loading and x,y seperate
path = 'https://bit.ly/advertising_csv'
adv = pd.read_csv(path)

target = 'Sales'
x = adv.drop(target, axis=1)
y = adv.loc[:, target]

#translate numpy and translate to tensor
x_np = x.values
y_np = y.values
x_tensor = torch.tensor(x_np, dtype=torch.float32)
y_tensor = torch.tensor(y_np, dtype=torch.float32)

device = 'cuda' if torch.cuda.is_available() else 'cpu'
print(f"Using {device} device.")
```

```python
#allocate the model to device(GPU)
model = nn.Sequential(nn.Linear(3, 1))
model = model.to(device)

#allocate data set to device
x,y = x_tensor.to(device), y_tensor.to(device)
```

```python
print(next(model.parameters()).device)
print(x.device, y.device)
```

```python
#GPU -> CPU
x_cpu = x.cpu()
print(x_cpu.device)
y_cpu = y.to("cpu")
print(y_cpu.device)

#CUP -> GPU
x_gpu = x_cpu.cuda()
print(x_gpu.device)
y_gpu = y_cpu.to("cuda")
print(y_gpu.device)
```
